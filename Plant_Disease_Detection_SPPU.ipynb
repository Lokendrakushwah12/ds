# Plant Disease Detection using CNN

This notebook demonstrates a Deep Learning practical for SPPU BE Computer Engineering 8th Semester.

The project:
- Loads plant disease dataset from local folders
- Preprocesses images
- Builds CNN model
- Trains model
- Tests predictions
- Displays graphs


````
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D
from tensorflow.keras.layers import Flatten, Dense, Dropout
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.preprocessing import image

import matplotlib.pyplot as plt
import numpy as np
````

## Dataset Folder Structure

Create folders like this:

```text
dataset/
    train/
        healthy/
        early_blight/
        late_blight/

    test/
        healthy/
        early_blight/
        late_blight/
```

Place leaf images inside corresponding folders.


````
# Dataset Preprocessing

train_datagen = ImageDataGenerator(
    rescale=1./255,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True
)

test_datagen = ImageDataGenerator(
    rescale=1./255
)

training_set = train_datagen.flow_from_directory(
    'dataset/train',
    target_size=(128,128),
    batch_size=32,
    class_mode='categorical'
)

test_set = test_datagen.flow_from_directory(
    'dataset/test',
    target_size=(128,128),
    batch_size=32,
    class_mode='categorical'
)
````

## CNN Architecture

The CNN model contains:
1. Convolution Layer
2. Pooling Layer
3. Flatten Layer
4. Dense Layer
5. Output Layer


````
# Build CNN Model

cnn = Sequential()

# First Convolution Layer
cnn.add(Conv2D(
    filters=32,
    kernel_size=3,
    activation='relu',
    input_shape=[128,128,3]
))

# First Pooling Layer
cnn.add(MaxPooling2D(pool_size=2, strides=2))

# Second Convolution Layer
cnn.add(Conv2D(
    filters=64,
    kernel_size=3,
    activation='relu'
))

# Second Pooling Layer
cnn.add(MaxPooling2D(pool_size=2, strides=2))

# Flattening
cnn.add(Flatten())

# Fully Connected Layer
cnn.add(Dense(units=128, activation='relu'))

# Dropout Layer
cnn.add(Dropout(0.5))

# Output Layer
cnn.add(Dense(units=3, activation='softmax'))
````

````
# Compile Model

cnn.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

cnn.summary()
````

````
# Train Model

history = cnn.fit(
    x=training_set,
    validation_data=test_set,
    epochs=10
)
````

````
# Save Model

cnn.save('plant_disease_model.h5')

print("Model Saved Successfully")
````

````
# Accuracy Graph

plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])

plt.title('Model Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')

plt.legend(['Training','Validation'])

plt.show()
````

````
# Loss Graph

plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])

plt.title('Model Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')

plt.legend(['Training','Validation'])

plt.show()
````

## Predict New Image

Place a test image in notebook folder:
```text
sample_leaf.jpg
```


````
# Predict New Image

test_image = image.load_img(
    'sample_leaf.jpg',
    target_size=(128,128)
)

test_image = image.img_to_array(test_image)
test_image = np.expand_dims(test_image, axis=0)

result = cnn.predict(test_image)

print(result)

print("Class Labels:")
print(training_set.class_indices)
````
